# Handle Qutlier

```python
Q1=df["Age"].quantile(0.25)
Q3=df["Age"].quantile(0.75)

IQR=Q3-Q1

lower_range=Q1-1.5*IQR
upper_range=Q3+1.5*IQR

outliers=[(df["Age"]<lower_range) & (df["Age"]>upper_range)]
print(outliers)
```

# EDA

```python
df.columns
```

Department-wise Gender Distribution

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x="Department",hue="Gender",palette=["#CC5500", "darkgreen"])
plt.title("Department-wise Gender Distribution",fontsize=11,fontweight="bold")
plt.xlabel("Department",fontsize=11,fontweight="bold")
plt.ylabel("Count of Empolyees",fontsize=11,fontweight="bold")
plt.xticks(rotation=45,fontsize=11)
plt.yticks(fontsize=11)
plt.legend(title="Gender",bbox_to_anchor=(1.02, 1),loc="upper left")
plt.grid(axis="y", linestyle="--", alpha=0.2)
plt.savefig("bar.png")
plt.show()
```

Current Salary by Department

```python
group=df.groupby("Department")["Current Salary"].sum()
print(group)
plt.figure(figsize=(10,6))
plt.bar(group.index,group,color="#CC5500")
plt.xticks(rotation=45,fontsize=11)
plt.yticks(fontsize=11)
plt.title("Total Current Salary by Department",fontsize=16,fontweight="bold")
plt.xlabel("Department",fontsize=12,fontweight="bold")
plt.ylabel("Total Current Salary(₹)",fontsize=12,fontweight="bold")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.savefig("bar9.png")
plt.show()
```

Previous Year Salary or Current Salary by Department

```python
plt.figure(figsize=(12,7))
sns.scatterplot(data=df,x="Previous Year Salary",y="Current Salary",hue="Department",style="Department",palette="Set2",marker="s",s=80,alpha=0.8)

plt.title("Previous Year Salary vs Current Salary by Department",fontsize=18,fontweight="bold")
plt.xlabel("Previous Year Salary (₹)", fontsize=13, fontweight="bold")
plt.ylabel("Current Salary (₹)", fontsize=13, fontweight="bold")

plt.legend(title="Department",bbox_to_anchor=(1.02, 1),loc="upper left")
plt.savefig("bar8.png")
plt.tight_layout()
plt.show()
```

Department wise Experience Distribution

```python
plt.figure(figsize=(18,9))
sns.countplot(data=df,x="Department",hue="Experience (Years)", palette="Set2")
plt.title("Department-wise Experience Distribution",fontsize=12,fontweight="bold")
plt.xlabel("Department",fontsize=12,fontweight="bold")
plt.ylabel("Experience (Years) count",fontsize=12,fontweight="bold")
plt.savefig("bar7.png")
plt.show()
```

Employee Count by City

```python
plt.figure(figsize=(15,6))
count_city=df.groupby("City")["Employee ID"].count()
sns.barplot(x=count_city.index,y=count_city,palette=["#CC5500", "darkgreen"])
plt.title("Employee Count by City",fontsize=14,fontweight="bold")
plt.xlabel("City",fontsize=12,fontweight="bold")
plt.ylabel("Employee Count",fontsize=12,fontweight="bold")
plt.xticks(fontsize=12,fontweight="bold")
plt.savefig("bar5.png")
plt.show()
```

Experience (Years) vs Current Salary by Gender

```python
plt.figure(figsize=(15,6))
sns.scatterplot(data=df,x="Experience (Years)",y="Current Salary",hue="Gender",palette="Set2",marker="s",s=80,alpha=0.8)
plt.title("Experience (Years) vs Current Salary by Gender",fontsize=16,fontweight="bold")
plt.xlabel("Experience (Years)",fontsize=12,fontweight="bold")
plt.ylabel("Current Salary (₹)",fontsize=12,fontweight="bold")
plt.legend(title="Gender",bbox_to_anchor=(1.02,1),loc="upper left")
plt.savefig("bar6.png")
plt.show()
```

Department wise Performance

```python
plt.figure(figsize=(14,7))
score_group=df.groupby("Department")["Current Year Score"].mean()
sns.barplot(x=score_group.index,y=score_group,palette="Set2")
plt.title("Department wise Performance",fontsize=16,fontweight="bold")
plt.xlabel("Department",fontsize=12,fontweight="bold")
plt.ylabel("Performance Score",fontsize=12,fontweight="bold")
plt.savefig("bar3.png")
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("RTA Dataset.csv.zip")
df
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Day_of_week",data=df,palette="Set2",hue="Day_of_week")
plt.title("Day of week")
plt.xlabel("Days")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
grouped_counts = df.groupby(['Age_band_of_driver', 'Sex_of_driver']).size().reset_index(name='Count')
grouped_counts
sns.barplot(x='Age_band_of_driver', y='Count', hue='Sex_of_driver', data=grouped_counts)
plt.title('Age Band vs Sex of Driver')
plt.xlabel('Age Band')
plt.ylabel('Count')
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Educational_level",data=df,palette="Set2")
plt.title("Educational Level")
plt.xlabel("Educational Level")
plt.ylabel("Count")
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Type_of_vehicle",data=df,palette="Set2")
plt.title("Type of Vehicle")
plt.xlabel("Type of Vehicle")
plt.ylabel("Count")
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Vehicle_movement",data=df,palette="Set2")
plt.title("Vehicle Movement")
plt.xlabel("Vehicle Movement")
plt.ylabel("Count")
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Cause_of_accident",data=df,palette="Set2")
plt.title("Cause of Accident")
plt.xlabel("Cause of Accident")
plt.ylabel("Count")
plt.xticks(rotation=45)
```
